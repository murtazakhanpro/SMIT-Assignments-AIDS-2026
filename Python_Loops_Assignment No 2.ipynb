# **Q8:** Ek list `numbers = [12, 45, 7, 23, 56, 90, 11]` mein se sabse bada (Maximum) number find karein loop ke zariye.

````python
# type your code here
numbers = [12, 45, 7, 23, 56, 90, 11]
greatest_num = numbers[0]

for n in (numbers):
    if n > greatest_num:
            greatest_num = n
            
print(greatest_num)
````

# **Q10:** 1 se 50 tak ke numbers check karein. Agar number 3 se divide hota hai to 'Fizz' print karein, agar 5 se hota hai to 'Buzz' print karein, aur agar dono se hota hai to 'FizzBuzz' print karein.

````python
# type your code here
# num = 0
for i in range(51):
    if i%3==0 or i%5==0:
        print("fizz-buzz" ,i, "divisible by 3 and 5")
    elif i%3==0:
        print("Fizz", i ,"divisible by 3")
    elif i%5==0:
        print("buzz", i, "Divisible by 5")
    
    else:
        print(i,"not Divisible",)
````

# **Q11:** Ek list di gayi hai `items = [1, 2, 3, 4, 5]`. Loop ka use karke ek nayi list banayein jis mein har number ka **Square** (e.g., 1, 4, 9...) ho.

````python
# type your code here
items = [1, 2, 3, 4, 5]
squared=[]
for i in (items):
    square_value = i**2
    squared.append(square_value)
    

print('items = ',items)
print("squared  = ",squared)
````
