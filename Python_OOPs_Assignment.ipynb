**Q2:** Aik `BankAccount` class banayein jismein `__balance` private variable ho. Aik `deposit()` aur aik `withdraw()` method banayein, check karein ke withdraw karte waqt balance zero se kam na ho.

```python
# Write your code here for Q2
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self , amount):
        self.__balance += amount

    def withdraw(self , withdraw):
        if withdraw <= self.__balance:
            self.__balance -= withdraw
            print("Sucessfull")
        else:
            print("Paisy Kam hain Account main")
        
```

**Q4:** Protected variable (`_`) aur Private variable (`__`) ke beech kya farq hota hai? Aik simple code ke zariye dono ka use-case dikhayein.

```python
# Write your code here for Q4
# protected variable main eik underscore hota ha (_)
# private variable main do underscore hote hain (__)

class answer:
    def __init__(self, protected , private):
        self._protected = protected
        self.__private = private

    def protected(self):
        print("yeh private variable hai")
    def private(self):
        print("yeh private variable hai")

obj = answer("Main Protected hoon", "Main Private hoon")

print(obj._protected)
print(obj.__private)
```

**Q6:** Single Inheritance ka use karte hue aik parent class `Vehicle` (with brand attribute) aur child class `Car` (with model attribute) banayein. Car ka object bana kar dono attributes print karein.

```python
# Write your code here for Q6
class Vehicle:
    def __init__(self, brand ):
        self.brand = brand
class car(Vehicle):
    def __init__(self, brand, model):
        super().__init__(brand) # Parent ke constructor ko chalaya
        self.model = model      # Apna model save kiya

my_car = car("Toyota", "Corolla")
print(my_car.brand)
print(my_car.model)
```

**Q7:** Multilevel Inheritance ka use karein. Class `Animal` -> Class `Mammal` -> Class `Dog`. Dog class ke object se parent classes ke features ko access karein.

```python
# Write your code here for Q7
class Animal:
    def eat(self):
          print("Khao")
class Mammal(Animal):
        def walk(self):
              print("chalo")

class Dog(Mammal):
      def bark(self):
            print("Barking...")


my_dog = Dog()
my_dog.eat()
my_dog.walk()
my_dog.bark()
```

**Q8:** Multiple Inheritance ka istemal karte hue do parent classes `Father` aur `Mother` banayein, aur unse aik child class `Child` derive karein jo dono ke habits inherit kare.

```python
# Write your code here for Q8
class Father:
    def Gardnening(self):
        print("Gardnening...")
class Mother:
    def Cooking(self):
        print("Cooking")
class child(Father , Mother):
    def Playing(self):
        print("Playing")

ObjChild = child()

ObjChild.Gardnening()
ObjChild.Cooking()
ObjChild.Playing()
```
